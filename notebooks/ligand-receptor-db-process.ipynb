```python
# =============================================================================
# Import libraries.
# =============================================================================

from pathlib import Path

import pandas as pd
```

```python
# =============================================================================
# Define project paths.
# =============================================================================

project_dir = Path.cwd().parents[0]

received_dir = project_dir / "data" / "received"
processed_dir = project_dir / "data" / "processed"
```

```python
# =============================================================================
# Load source workbook sheets.
# =============================================================================

source_file = received_dir / "Ligand_Receptor_DB.xlsx"

ligand_receptor = pd.read_excel(
    source_file,
    sheet_name = "Ligand-Receptor",
)

genes = pd.read_excel(
    source_file,
    sheet_name = "genes",
)
```

```python
# =============================================================================
# Export analysis-ready files.
# =============================================================================

ligand_receptor.to_csv(
    processed_dir / "ligand_receptor.csv",
    index = False,
)

ligand_receptor.to_parquet(
    processed_dir / "ligand_receptor.parquet",
    index = False,
)

genes.to_csv(
    processed_dir / "genes.csv",
    index = False,
)
```
